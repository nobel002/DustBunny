```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

PATH = "../target/release/Energy_Latice.csv"

data = pd.read_csv(PATH, sep="\t")

t, E = data["t"], data["energy"]
plt.plot(t, E)
plt.show()
print((max(E)-min(E))/max(E))

dt = t[1] - t[0]
signal = E - np.mean(E)

fft = np.fft.rfft(signal)
freq = np.fft.rfftfreq(len(signal), dt)

power = 2*np.log(np.abs(fft))

mask = freq < 25
mask[0] = False
plt.plot(freq[mask], power[mask])
plt.xlabel("Frequency")
plt.ylabel("Power")
plt.title("Hamiltonian drift spectrum")
plt.show()
```
